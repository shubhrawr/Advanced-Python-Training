accept 5 marks, roll no, name and calculate total & percentage if percebtage >=85 student can take admission else cant take admission using class & object & constructor


```python
class Student:
  def __init__(self,roll_no,name,marks):
    self.roll_no = roll_no
    self.name = name
    self.marks = marks
  def show_result(self):
        total = sum(self.marks)
        percentage = total / 500 * 100

        print(f"\nRoll number: {self.roll_no}")
        print(f"Name: {self.name}")
        print(f"Total marks: {total}/500")
        print(f"Percentage: {percentage:.2f}%")

        if percentage >= 85:
            print("Can take admission.")
        else:
            print("Cannot take admission.")


roll_no = input("Enter roll number: ")
name = input("Enter name: ")

marks = []
for i in range(5):
    mark = float(input(f"Enter marks for subject {i + 1}: "))
    marks.append(mark)

s = Student(roll_no, name, marks)
s.show_result()
```

```python
class A:#single level inheritance
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")
obj=B()
obj.showB()
obj.showA()
```

```python
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")
class C(B):
  def showC(self):
    print("showC function")
obj=C()
obj.showB()
obj.showA()
obj.showC()
```

```python
class A:#multiple inheritance
  def showA(self):
    print("showA function")

class B:
  def showB(self):
    print("showB function")
class C(A,B):
  def showC(self):
    print("showC function")
obj=C()
obj.showB()
obj.showA()
obj.showC()
```

```python
#hierarchal inheritance
class D:
  def show(self):
    print("showD function")
class A(D):
  def showA(self):
    print("showA function")

class B(D):
  def showB(self):
    print("showB function")
class C(D):
  def showC(self):
    print("showC function")
obj1=A()
obj1.showA()
obj1.show()

obj2=B()
obj2.showB()
obj2.show()

obj3=C()
obj3.showC()
obj3.show()
```
