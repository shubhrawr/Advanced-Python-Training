```python
from abc import ABC, abstractmethod
class Test(ABC):
  @abstractmethod
  def m1(self):
    pass
  @abstractmethod
  def m2(self):
    pass
class MyPython(Test):
  pass
class MyPython(Test):
  def m1(self):
    print("m1 implementation")
  @abstractmethod
  def m2(self):
    pass
class Child(MyPython):
  def m2(self):
    print("m2 implementation")
m=Child()
m.m1()
m.m2()
```

```python
class Test1:
  _b = None
  __c = None
  def __init__(self):
    self.a = 101
    self._b = 201
    self.__c = 301

  def show1(self):
    print("a=", self.a)
    print("b=", self._b)
    print("c=", self.__c)

class Test2(Test1):
  def show2(self):
    print("a=", self.a)
    print("b=", self._b)
    print("c=", self._Test1__c)

class Test3(Test2):
  def show3(self):
    print("a=", self.a)
    print("b=", self._b)
    print("c=", self._Test1__c)

t1 = Test1()
t1.show1()
t2 = Test2()
t2.show2()
t3 = Test3()
t3.show3()
```

```python
import os, sys
fname=input("Enter file name: ")
if os.path.isfile(fname):
  print("File exist: ",fname)
  f=open(fname,'r')
else:
  print("File doesn't exist: ",fname)
  sys.exit(0)

lcount=wcount=ccount=0
for line in f:
  lcount=lcount+1
  ccount=ccount+len(line)
  words=line.split()
  wcount=wcount+len(words)

print("Total lines: ",lcount)
print("Total words: ",wcount)
print("Total characters: ",ccount)
```
